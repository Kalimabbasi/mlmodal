```python
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer

# Example corpus: replace this with your actual documents list or data loader
documents = [
    "Natural language processing and machine learning are exciting fields.",
    "TF-IDF is a numerical statistic that reflects how important a word is to a document.",
    "Machine learning models require clean data and robust feature engineering.",
    "Python is a popular programming language for data science and artificial intelligence."
]

# 1. Initialize the TF-IDF vectorizer (removes standard English stop words)
vectorizer = TfidfVectorizer(stop_words='english')

# 2. Fit and transform the documents into a TF-IDF matrix
tfidf_matrix = vectorizer.fit_transform(documents)

# 3. Retrieve the feature names (vocabulary words)
feature_names = vectorizer.get_feature_names_out()

# 4. Extract and print the top-5 terms for each document
top_n = 5
for i, doc in enumerate(documents):
    # Get the sparse/dense row for the current document
    row = tfidf_matrix.toarray()[i]

    # Sort indices by TF-IDF score in descending order
    sorted_indices = np.argsort(row)[::-1]

    # Filter out terms with a score of 0 and take the top N
    top_terms = [(feature_names[idx], row[idx]) for idx in sorted_indices[:top_n] if row[idx] > 0]

    print(f"Document {i + 1} Summary (Top Terms):")
    for term, score in top_terms:
        print(f"  - {term}: {score:.4f}")
    print("-" * 40)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.feature_extraction.text import TfidfVectorizer

def rank_documents_by_query(query, vectorizer, tfidf_matrix, documents):
    """
    Transforms a query using the fitted vectorizer, computes cosine similarity
    against all documents, and ranks them in descending order.
    """
    # 1. Transform the query using the already-fitted vectorizer
    query_vector = vectorizer.transform([query])

    # 2. Compute cosine similarity between the query vector and all document vectors
    # cosine_similarity returns a 2D array, so we flatten it to 1D
    similarity_scores = cosine_similarity(query_vector, tfidf_matrix).flatten()

    # 3. Get indices sorted by similarity score in descending order
    ranked_indices = similarity_scores.argsort()[::-1]

    # 4. Compile the ranked results
    ranked_results = []
    for rank, idx in enumerate(ranked_indices, start=1):
        ranked_results.append({
            "rank": rank,
            "doc_id": idx + 1,
            "score": similarity_scores[idx],
            "text": documents[idx]
        })

    return ranked_results

# --- Complete Example with 10 Documents ---
documents = [
    "Natural language processing and machine learning are exciting fields.",
    "TF-IDF is a numerical statistic that reflects how important a word is to a document.",
    "Machine learning models require clean data and robust feature engineering.",
    "Python is a popular programming language for data science and artificial intelligence.",
    "Deep learning uses neural networks with multiple layers to learn complex patterns.",
    "Information retrieval systems help users find relevant documents from a large corpus.",
    "Data visualization helps communicate complex insights through charts and graphs.",
    "Supervised learning algorithms are trained using labeled input and output data.",
    "Cloud computing provides scalable infrastructure for deploying modern software applications.",
    "Natural language understanding enables computers to comprehend human speech and text."
]

# Fit vectorizer and build TF-IDF matrix
vectorizer = TfidfVectorizer(stop_words='english')
tfidf_matrix = vectorizer.fit_transform(documents)

# Test the function with a query
query = "machine learning models"
results = rank_documents_by_query(query, vectorizer, tfidf_matrix, documents)

print(f"Query: '{query}'\n")
for res in results:
    print(f"Rank {res['rank']} | Doc #{res['doc_id']} | Score: {res['score']:.4f}")
    print(f"   -> {res['text']}\n")
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

def rank_documents_by_query(query, vectorizer, tfidf_matrix, documents):
    """
    Transforms a query using the fitted vectorizer, computes cosine similarity
    against all documents, and ranks them in descending order.
    """
    query_vector = vectorizer.transform([query])
    similarity_scores = cosine_similarity(query_vector, tfidf_matrix).flatten()
    ranked_indices = similarity_scores.argsort()[::-1]

    ranked_results = []
    for rank, idx in enumerate(ranked_indices, start=1):
        ranked_results.append({
            "rank": rank,
            "doc_id": idx + 1,
            "score": similarity_scores[idx],
            "text": documents[idx]
        })
    return ranked_results

# 1. Define a 10-document corpus containing the target categories
documents = [
    # Document 1 (Target for Climate Policy query)
    "Climate Change: renewable energy and climate policy are essential to reduce global carbon emissions and mitigate environmental damage.",
    # Document 2 (Target for Security query)
    "Cybersecurity: password security and phishing attacks are major threats to modern digital infrastructure and user privacy.",
    # Document 3 (Target for Health query)
    "Public Health: healthy sleep exercise and diet play a critical role in maintaining long-term physical and mental wellness.",
    # Additional documents (4-10)
    "Artificial Intelligence: machine learning models require clean data and robust feature engineering.",
    "Data Science: Python is a popular programming language for data analysis and artificial intelligence.",
    "Cloud Computing: cloud infrastructure provides scalable deployment for modern enterprise software.",
    "Economics: inflation rates and monetary policy strongly influence global financial market stability.",
    "Astronomy: telescopes and space probes help scientists discover distant exoplanets and galaxies.",
    "Education: interactive learning platforms improve student engagement and knowledge retention.",
    "Transportation: electric vehicles and public transit networks reduce urban traffic congestion."
]

# 2. Fit vectorizer and build the TF-IDF matrix
vectorizer = TfidfVectorizer(stop_words='english')
tfidf_matrix = vectorizer.fit_transform(documents)

# 3. Define the three predefined queries
queries = [
    "renewable energy and climate policy",
    "password security and phishing",
    "healthy sleep exercise and diet"
]

# 4. Run and observe the highest-ranking documents for each query
for query in queries:
    print(f"=" * 60)
    print(f"Query: '{query}'")
    print(f"=" * 60)

    results = rank_documents_by_query(query, vectorizer, tfidf_matrix, documents)

    # Print the top-1 (or top-3) results to observe ranking
    for res in results[:3]:
        print(f"Rank {res['rank']} | Score: {res['score']:.4f}")
        print(f"   -> {res['text']}\n")
```
